title: "E) Feedforward"

Trajectory planning is a technique in which you simplify the trajectory which ensures that the error gets lower. Most often uses more differentiable signals. This leads to:
- less oscillations
- possibly faster settling times
- smaller control signals

The DC-gain is the steady-state ain of a system where $\omega$ is 0. Often this is $\frac{1}{K_p}

Feedforward is added to after the controller. Mostly the feedforward should actually be the inverse of the plant. However if the plant is unstable or unknown then we need to approximate the inverse of the plant in the time domain. Generally you forward the forces that are needed to: 
- accelerate the rigid body mode of the plant $F_a = K_{fa}\ddot x_s$
- overcome the viscous friction with the fixed world: $F_v = K_{fv} \dot x_s$
- overcome the coulomb friction with the fixed world: $F_c = K_{fc} \cdot sign(\dot x_s)$


## Tuning of Feedforward

The best trajectory to tune the parameters is approx 1/3 acceleration, 1/3 constant velocity, 1/3 deceleration.

First the $K_{fc}$ should be tuned. In such a way that the deceleration of the forward movement and acceleration of the backward movement have the same error.

Secondly the $K_{fv}$ must be tunes. There should be no error during constant velocity phases when $K_{fc}$ and $K_{fv}$ are correct.

Thirdly $K_{fa}$ must be tuned to minize the error during acceleration and deceleration.

For this tuning the feedback controller should be just good enough to reasonably control the plant but should be not too good such that the error deminishes. The frequency of the reference signal should be such that the magnitude of the closed loop sensitivity is around -40.  Or the bandwidth $f_{bw} \ge \frac{2}{t_{phase}} \approx 20 hz $ where $t_{phase}$ is the length of each phase.
