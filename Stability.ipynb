When using a feedback controller the poles become the zeros. However everytime checking how these poles changes based on the controller is not really feasible or intuitive, that is why nyquist criterion is designed

This criterion is based on encirclements the number of clockwise encirclements of the origin is denoted by $N=Z-P$ in which Z is the number of enclosed zeors and P is the number of enclosed poles. this is evaluated on 1 + L(s) because:

- Poles of 1 + L(s) -> open-loop poles
- Zeros of 1 + L(s) -> closed-loop poles

The closed loop is stable when $Z$ is 0 and therefore $Z=N+P=0$

However this can be simplified to instead of 1+L(s) encirclements of origin to just L(s) encirclements of point(-1,0)

This can be translated to the bode plot and therefore a bode plot (L(s)) has the following necessary conditions for stability:
- Slope $\approx -1 at $w_{co}$
- the phase  must be between -180 and 0, therefore the curve meest keep (-1,0) to its left. 
- It must enter the unit disc to its left

However the following sufficient conditions must also hold:

- All poles and zeros of $L(S)=C(s)H(s)$ must lie strictly in the LHP
- Single crossover. The magnitude crosses 0dB once and never crosses it again. 

## Stability Margins
A system is stable or not however it does have some robustness measurements. These are also called phase margin and gain margin. They are also provided in the plot below.

The PM is the phase lead above -180 at the frequency where $|L(s)| = 0dB$,

The inverse of GM is the gain at the frequency where $\angle L(j\omega) = -180^\circ$

The modulus margin (MM) is the inverse of the smallest distance of $L(j\omega)$ over all $\omega$ to the point -1. The distance of the point -1 to $L(j\omega)$ is given by: 
$$L(j\omega) - (-1) = 1 + L(j\omega)$$

Note that the modulus margin is therefore the minimal value of this function. Also know that the Sensitivity is given as:
$$S(j\omega) = \frac{1}{1 + L(j\omega)}$$
Therefore sensitivity is the inverse of the distance and therefore the minimal value of the distance function is the same as the maximal value of the sensitivity function.
