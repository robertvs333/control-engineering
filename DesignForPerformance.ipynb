Normally in feedback $e=r-y$ rewriting this gives:
\begin{equation}
\frac{e}{r} = \frac{1}{1+HC} = S
\end{equation}

when using this y can be written as:
\begin{equation}
y = \frac{H}{1+HC}d=S\cdot Hd
\end{equation}

So therefore feedback is:
- **advantageous** when $|S|<1$
- **disadvantageous** when $|S| > 1$

However due to the bode sensitivity integral you cannot just ensure that $|S|$ is always lower than 1. Becuase the area beneath $|S|$ is always non-negative. Therefore shaping $|S|$ is like a waterbed suppression for 1 frequency leads to a amplification elsewhere. This holds only where the open loops $L(s)$ with relative degree 2 or higher. 

## Closed loop transfer functions
Some standard closed loop transfer function are derived based on the image provided above.

**Sensitivity:**
\begin{equation}
S=\frac{e}{r} = \frac{u}{d} = \frac{1}{1+HC}
\end{equation}

**Complementary sensitivity:**
\begin{equation}
T=\frac{y}{r} = \frac{y}{\eta} = \frac{HC}{1+HC}
\end{equation}

**Process sensitivity:**
\begin{equation}
HS = \frac{y}{d}=\frac{H}{1+HC}
\end{equation}

**Control Sensitivity:**
\begin{equation}
    CS=\frac{u}{r}=\frac{C}{1+HC}
\end{equation}

So the general goal is to minimize the error $e=0$ -> minimizing $SS and tracking the input $y=r$ -> $T\approx1$. This is both achieved when $HC$ is maximized. this is called high-gain feedback. 

## Bandwitdh

Most often bandwidth should be chosen as high as possible but not more than necessary. This is also stated the other way around as: "Choose the bandwidth as low as possible, but such that the closed loop still meets its requirements"
